```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = input("Enter a sentence")
```

```python
doc = nlp(text)
```

```python
if len(doc.ents) == 0:
  print("\nNo Named Entity Found.")
  print("Please senter a sentence containing a person's name organisation, place, date and income etc.")
else:
  print("\nNamed Entities Found:")
  for ent in doc.ents:
    print(ent.text, ent.label_)
```
