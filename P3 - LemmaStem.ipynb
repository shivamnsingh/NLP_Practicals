```python
import nltk
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
text = " My name is Eren Yeager... The hardening of all the walls... has been undone, and the Titans buried within them have started to march. My goal is to protect the people of Paradis... But the world desires the extinction of our people... I reject their desire"
tokens = word_tokenize(text)
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
print("stemming:")
for word in tokens:
    print(f"{word} -> {stemmer.stem(word)}")
```

```python
print("\nLemmatization:")
for word in tokens:

    print(f"{word} -> {lemmatizer.lemmatize(word)}")
```
